1071. Greatest Common Divisor of Strings

For two strings s and t, we say "t divides s" if and only if s = t + t + t + ... + t + t (i.e., t is concatenated with itself one or more times).

Given two strings str1 and str2, return the largest string x such that x divides both str1 and str2.

 

Example 1:

Input: str1 = "ABCABC", str2 = "ABC"

Output: "ABC"

Example 2:

Input: str1 = "ABABAB", str2 = "ABAB"

Output: "AB"

Example 3:

Input: str1 = "LEET", str2 = "CODE"

Output: ""

Example 4:

Input: str1 = "AAAAAB", str2 = "AAA"

Output: ""​​​​​​​

 

Constraints:

    1 <= str1.length, str2.length <= 1000
    str1 and str2 consist of English uppercase letters.



```python
class Solution:
    def gcdOfStrings(self, str1: str, str2: str) -> str:
        len1, len2 = len(str1), len(str2)

        result = ""
        # loop from large to small values. Start with the length of the shorter string
        for i in range(min(len1, len2), 0, -1):
            if len1 % i != 0 or len2 % i != 0:
                continue
            # the multiple for the common divisor. If i is 2, and len1 is 5, that means the divisor would need to repeat 5 times to equal str1
            n1, n2 = len1 // i, len2 // i
            # common divisor
            cd = str1[:i]
            if str1 == n1 * cd and str2 == n2 * cd:
                result = cd
                break
        return result
```

```python
solution = Solution()
test_cases = [
    ("ABCABC", "ABC", "ABC"),
    ("ABABAB", "ABAB", "AB"),
    ("LEET", "CODE", ""),
    ("AAAAAA", "AAA", "AAA"),
    ("asdffasdffasdffasdff", "asdffasdffasdff", "asdff"),
]
for str1, str2, expected in test_cases:
    result = solution.gcdOfStrings(str1, str2)
    assert result == expected, f"Test case failed for {str1}, {str2}. Expected {expected}, got {result}"

print("All test cases passed!")
```
