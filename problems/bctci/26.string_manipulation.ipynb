# String Manipulation Problems

## 26.1: Is alphanumeric

Implement a function that returns whether a character is a lowercase or uppercase english letter or a digit between '0' and '9'

```python
def is_alphanumeric(c):
    if not (isinstance(c, str) and len(c)==1):
        return False
    if ord(c) >= ord('A') and ord(c) <= ord('Z'):
        return True
    if ord(c) >= ord('a') and ord(c) <= ord('z'):
        return True
    if ord(c) >= ord('0') and ord(c) <= ord('9'):
        return True
    return False
```

```python
print(is_alphanumeric('a'))
print(is_alphanumeric('z'))
print(is_alphanumeric('g'))
print(is_alphanumeric('G'))
print(is_alphanumeric('5'))
print(is_alphanumeric('0'))
print(is_alphanumeric('a'))
print(is_alphanumeric(5))
print(is_alphanumeric(1000))
print(is_alphanumeric('aasfdg'))
```
