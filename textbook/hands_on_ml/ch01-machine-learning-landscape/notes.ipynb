# Machine Learning Landscape

This chapter is a high level overview of machine learning and its applications in the real world and the theory behind it. 

## Examples of Applications:

Image classification is typically performed using Convolutional Neural Networks (CNNs) or Visual Transformers. Same with semantic image segmentation.

Natural Language Processing (NLP) is often tackled using recurrent neural networks (RNNs) or transformers.

Estimating Genetic Risk to Disease often uses State Space Models.

## Types of Machine Learning Systems:

Systems can be organized by multiple criteria, including 
- Learning Paradigm: Supervised, Unsupervised, Semi-supervised, and Reinforcement Learning.
- Learning Type: Online vs Batch
- Whether they simply compare new data to known data or whether they build a model of the underlying data distribution. (Instance-based vs Model-based)

### Training Supervision

- Supervised: training set has labels / targets
- Unsupervised: training data is unlabeled
- Semi-supervised: training data is partially labeled
- Self-supervised: training data is unlabeled but the model generates its own labels
- Reinforcement Learning: the model learns by interacting with an environment and receiving feedback in the form of rewards or penalties

```python
# Training and running a linear model with scikit-learn
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.neighbors import KNeighborsRegressor
```

```python
# download and prepare data
data_root = "https://github.com/ageron/data/raw/main/"
lifesat = pd.read_csv(data_root + "lifesat/lifesat.csv")
X = lifesat[["GDP per capita (USD)"]].values
Y = lifesat[["Life satisfaction"]].values

# Visualize data
lifesat.plot(kind="scatter", grid=True, x="GDP per capita (USD)", y="Life satisfaction")
plt.axis([23_500, 62_500, 4, 9])
plt.show()
```

```python
# use linear regression to predict Puerto Rico's Life Sat. This is a model based approach
model = LinearRegression()

model.fit(X, Y)

X_new = [[33_442.8]]
print(model.predict(X_new))
```

```python
# Use K nearest neighbors to predict Puerto Rico's Life Sat. This is an instance based approach
model = KNeighborsRegressor(n_neighbors=3)

model.fit(X, Y)

X_new = [[33_442.8]]
print(model.predict(X_new))
```
